Este cuaderno hace **lo mismo que acabas de hacer en Orange**: agrupar las flores de Iris sin decirle al computador de qué especie es cada una.

```python
# 1. Traer las herramientas (se ejecuta una sola vez)
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
print("Listo: herramientas cargadas.")
```

## 1. Mirar los datos

Es la misma tabla que abriste en Orange con **Data Table**: 150 flores, cuatro medidas por flor y la especie.

```python
# 2. Cargar la tabla de flores Iris
iris = load_iris(as_frame=True)
flores = iris.frame
flores["especie"] = flores["target"].map(dict(enumerate(iris.target_names)))
flores = flores.drop(columns="target")
print("Filas:", len(flores), "| Columnas:", len(flores.columns))
flores.head()
```

## 2. Dibujar las flores

Igual que el **Scatter Plot** de Orange: cada punto es una flor. Aquí se colorea por su especie **real**, para tener con qué comparar después.

```python
# 3. Grafico coloreado por la especie REAL
x, y = "petal length (cm)", "petal width (cm)"
for especie, grupo in flores.groupby("especie"):
    plt.scatter(grupo[x], grupo[y], label=especie)
plt.xlabel("largo del pétalo (cm)")
plt.ylabel("ancho del pétalo (cm)")
plt.title("Las flores, coloreadas por su especie real")
plt.legend()
plt.show()
```

## 3. Agrupar sin decirle la especie

Ahora le quitamos la respuesta al computador: solo recibe las cuatro medidas y debe formar **k** grupos. Es K-Means, el mismo que trazaste a mano en la clase.

Empieza con **k = 3** y ejecuta la celda.

```python
# 4. Agrupar con K-Means
k = 3   # ⬅️ CAMBIA ESTE NÚMERO: prueba 2, 3, 4 y 5

medidas = flores.drop(columns=["especie", "grupo"], errors="ignore")  # sin la respuesta
modelo = KMeans(n_clusters=k, n_init=10, random_state=0)
flores["grupo"] = modelo.fit_predict(medidas)

for g, grupo in flores.groupby("grupo"):
    plt.scatter(grupo[x], grupo[y], label=f"grupo {g}")
plt.xlabel("largo del pétalo (cm)")
plt.ylabel("ancho del pétalo (cm)")
plt.title(f"Los {k} grupos que encontró el computador")
plt.legend()
plt.show()
```

## 4. ¿Se parecen los grupos a las especies?

Esta tabla cruza los grupos que encontró el computador con las especies reales. Cada número dice cuántas flores de esa especie quedaron en ese grupo.

```python
# 5. Comparar los grupos contra las especies reales
pd.crosstab(flores["grupo"], flores["especie"])
```
