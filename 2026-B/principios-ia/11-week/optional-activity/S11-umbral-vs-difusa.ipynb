## Paso 0 · Preparar el cuaderno

Esta celda carga las herramientas para hacer tablas y gráficas, y escribe los datos de las dos tablas del Excel. Ejecútela una vez y siga.

````python
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams["figure.dpi"] = 100
plt.rcParams["font.size"] = 11

VERDE, ROJO, NARANJA, GRIS = "#2e9e44", "#d64545", "#e07b39", "#9aa0a6"

# Tabla 1 del Excel (verde)
materias = ["Principios de IA", "Lógica Programación", "Lógica Matemática",
            "Cálculo", "Introducción a la Ingeniería"]
notas = [3.0, 4.9, 3.6, 2.99, 4.2]

# Tabla 2 del Excel (azul)
personas = ["Juan", "Pedro", "María", "Diego", "Wendy"]
estaturas = [175, 185, 190, 210, 150]

print("Listo. Ya puede seguir con el paso 1.")
````

# PARTE A · Tabla 1: las notas

## Paso 1 · La regla de siempre

Usted ya conoce esta regla:

> **Con 3,0 o más, pasó. Con menos de 3,0, perdió.**

Ese **3,0** se llama **umbral**. Es como una **raya pintada en el piso**: de un lado está «pasó» y del otro lado está «perdió». No hay un punto medio.

Ejecute la celda para ver la tabla con el resultado de cada materia.

````python
UMBRAL_NOTA = 3.0
resultado = ["Pasó" if n >= UMBRAL_NOTA else "Perdió" for n in notas]

tabla1 = pd.DataFrame({"Materia": materias, "Nota": notas, "Resultado": resultado})
tabla1
````

## Paso 2 · Veamos la raya en una gráfica

Cada barra es una materia. La **línea punteada** es el umbral (3,0).

- Las barras **verdes** llegan a la raya o la pasan: **pasó**.
- La barra **roja** se queda corta: **perdió**.

````python
colores = [VERDE if n >= UMBRAL_NOTA else ROJO for n in notas]

plt.figure(figsize=(9, 4))
barras = plt.barh(materias, notas, color=colores)
plt.axvline(UMBRAL_NOTA, color="black", linestyle="--", linewidth=2, label="umbral 3,0")
plt.legend(loc="lower right")
for b, n, r in zip(barras, notas, resultado):
    plt.text(n + 0.05, b.get_y() + b.get_height() / 2, f"{n}  {r}", va="center")
plt.xlim(0, 6)
plt.xlabel("Nota")
plt.title("Gráfica 1 · Cada materia contra la raya del 3,0")
plt.gca().invert_yaxis()
plt.show()
````

## Paso 3 · El detalle que importa: 3,0 contra 2,99

Fíjese en dos materias:

- **Principios de IA:** 3,0 → **Pasó**.
- **Cálculo:** 2,99 → **Perdió**.

La diferencia es **una centésima**. Si las pone una al lado de la otra, casi no se distinguen. Pero el resultado es **totalmente opuesto**.

La gráfica hace un **zoom** cerca del 3,0 para que lo vea.

````python
plt.figure(figsize=(8, 3.2))
plt.barh(["Cálculo", "Principios de IA"], [2.99, 3.0], color=[ROJO, VERDE], height=0.5)
plt.axvline(UMBRAL_NOTA, color="black", linestyle="--", linewidth=2)
plt.text(2.9805, 0, " 2,99 → Perdió", va="center", color="white", fontweight="bold")
plt.text(2.9805, 1, " 3,00 → Pasó", va="center", color="white", fontweight="bold")
plt.xlim(2.98, 3.005)
plt.xlabel("Nota (zoom entre 2,98 y 3,00)")
plt.title("Gráfica 2 · Una centésima de diferencia, resultados opuestos")
plt.show()
````

## Paso 4 · El «escalón»

Si dibujamos la regla para **todas** las notas posibles, de 0 a 5, sale un **escalón**:

- Antes del 3,0 la respuesta vale **0** (perdió).
- Desde el 3,0 la respuesta salta a **1** (pasó).

No hay rampa ni subida suave: es un **salto**. A esto se le llama **lógica clásica**: solo existen el 0 y el 1.

````python
xs = [i / 100 for i in range(0, 501)]
ys = [1 if x >= UMBRAL_NOTA else 0 for x in xs]

plt.figure(figsize=(9, 3.5))
plt.plot(xs, ys, color="black", linewidth=3)
for m, n in zip(materias, notas):
    y = 1 if n >= UMBRAL_NOTA else 0
    plt.scatter(n, y, s=90, color=VERDE if y else ROJO, zorder=3)
plt.annotate("aquí está el salto", xy=(3.0, 0.5), xytext=(0.6, 0.55),
             arrowprops=dict(arrowstyle="->"), fontsize=11)
plt.yticks([0, 1], ["0 = Perdió", "1 = Pasó"])
plt.xlabel("Nota")
plt.ylim(-0.3, 1.3)
plt.title("Gráfica 3 · La lógica clásica dibujada: un escalón")
plt.grid(alpha=0.3)
plt.show()
````

### Conclusión de la parte A

Con las notas, el escalón **está bien**: el reglamento necesita una respuesta de sí o no, porque la materia se pasa o se pierde.

Pero hay palabras que **no son de sí o no**, como «alto», «caliente» o «lleno». Ahí el escalón se ve injusto. Eso es lo que muestra la tabla 2.

# PARTE B · Tabla 2: la estatura

## Paso 5 · Las cinco personas

La pregunta ahora es: **¿esta persona es alta?**

Primero miremos a las cinco personas ordenadas de la más baja a la más alta.

````python
orden = sorted(zip(estaturas, personas))
plt.figure(figsize=(8, 4))
plt.bar([p for _, p in orden], [e for e, _ in orden], color="#5aa9e6")
for i, (e, p) in enumerate(orden):
    plt.text(i, e + 2, f"{e} cm", ha="center")
plt.ylim(0, 230)
plt.ylabel("Estatura (cm)")
plt.title("Gráfica 4 · Las cinco personas, de la más baja a la más alta")
plt.show()
````

## Paso 6 · Respuesta con lógica clásica (umbral de 180 cm)

Hacemos lo mismo que con las notas: pintamos una raya en **180 cm**.

> **Si mide 180 cm o más → 1 (es alta). Si mide menos → 0 (no es alta).**

Es el «**A: 180**» del Excel.

````python
UMBRAL_ALTO = 180
clasica = [1 if e >= UMBRAL_ALTO else 0 for e in estaturas]

plt.figure(figsize=(8, 4))
plt.bar([p for _, p in orden], [e for e, _ in orden],
        color=[VERDE if e >= UMBRAL_ALTO else ROJO for e, _ in orden])
plt.axhline(UMBRAL_ALTO, color="black", linestyle="--", linewidth=2)
plt.text(4.45, UMBRAL_ALTO + 14, "umbral 180 cm", fontsize=10, ha="right")
for i, (e, p) in enumerate(orden):
    plt.text(i, e / 2, ("1\nalta" if e >= UMBRAL_ALTO else "0\nno alta") + f"\n{e} cm",
             ha="center", va="center", color="white", fontweight="bold")
plt.ylim(0, 235)
plt.ylabel("Estatura (cm)")
plt.title("Gráfica 5 · Lógica clásica: o es alta (1) o no lo es (0)")
plt.show()

pd.DataFrame({"Persona": personas, "Estatura (cm)": estaturas, "Clásica (A: 180)": clasica})
````

## Paso 7 · ¿Qué tiene de raro?

Mire con calma estos dos casos:

1. **Juan (175)** y **Pedro (185)** se llevan **10 cm**. Pero la regla dice que Juan es **0** y Pedro es **1**: extremos opuestos.
2. **Pedro (185)** y **Diego (210)** se llevan **25 cm**. Pero la regla dice que los dos son **1**: «igual de altos».

Es decir: la regla separa a los que son **parecidos** y junta a los que son **distintos**. Eso choca con el sentido común.

````python
fig, ejes = plt.subplots(1, 2, figsize=(10, 4), sharey=True)

ejes[0].bar(["Juan", "Pedro"], [175, 185], color=[ROJO, VERDE])
ejes[0].set_title("Casi iguales (10 cm)…\n…pero 0 y 1")
ejes[1].bar(["Pedro", "Diego"], [185, 210], color=[VERDE, VERDE])
ejes[1].set_title("Muy distintos (25 cm)…\n…pero los dos 1")
for ax, vals in zip(ejes, [[175, 185], [185, 210]]):
    ax.axhline(UMBRAL_ALTO, color="black", linestyle="--")
    for i, v in enumerate(vals):
        ax.text(i, v / 2, f"{v} cm", ha="center", color="white", fontweight="bold")
    ax.set_ylim(0, 230)
ejes[0].set_ylabel("Estatura (cm)")
fig.suptitle("Gráfica 6 · Lo que no cuadra con el umbral de 180", fontweight="bold")
plt.tight_layout()
plt.show()
````

Por eso en el Excel marcamos a **Juan, Pedro y María** como «**0 y 1**»: no son ni del todo bajos ni del todo altos. Están **en el medio**.

## Paso 8 · La idea de la lógica difusa: una batería de 0 a 1

La lógica difusa cambia la pregunta:

| Lógica clásica | Lógica difusa |
|---|---|
| ¿Es alta? **Sí o no** | ¿**Qué tan** alta es? **De 0 a 1** |

Piense en la **batería del celular**:

- **0** = batería vacía → «nada alto».
- **1** = batería llena → «alto del todo».
- **0,5** = media batería → «alto a medias».

A ese número entre 0 y 1 se le llama **grado de pertenencia**. Es lo que en el Excel quedó como «**0, 0 y 1, 1**»: 0 para quien claramente no es alto, un valor **entre 0 y 1** para el medio, y 1 para quien claramente sí lo es.

## Paso 9 · La rampa: cómo se llena la batería

Para saber cuánta «batería de alto» tiene cada persona usamos una **rampa**, como la de un parqueadero:

- **Abajo de la rampa (hasta 160 cm):** batería en **0**.
- **Arriba de la rampa (desde 200 cm):** batería en **1**.
- **Subiendo la rampa (entre 160 y 200 cm):** la batería se va llenando **poco a poco**.

Los números 160 y 200 **los elige quien diseña** el sistema. Aquí los escogimos para que Juan, Pedro y María queden en el medio, como en el Excel.

````python
COMIENZO, FINAL = 160, 200

def grado(cm, comienzo=COMIENZO, final=FINAL):
    if cm <= comienzo:
        return 0.0
    if cm >= final:
        return 1.0
    return (cm - comienzo) / (final - comienzo)

xs = list(range(140, 221))
plt.figure(figsize=(9, 4))
plt.axvspan(140, COMIENZO, color=ROJO, alpha=0.12)
plt.axvspan(COMIENZO, FINAL, color=NARANJA, alpha=0.12)
plt.axvspan(FINAL, 220, color=VERDE, alpha=0.12)
plt.plot(xs, [grado(x) for x in xs], color=NARANJA, linewidth=4)
plt.text(143, 0.85, "ZONA 1\nno es alto\n(grado 0)", fontsize=10)
plt.text(170, 0.85, "ZONA 2\nla rampa: «un poco alto»\n(grado entre 0 y 1)", fontsize=10)
plt.text(202, 0.45, "ZONA 3\nalto del todo\n(grado 1)", fontsize=10)
plt.xlabel("Estatura (cm)")
plt.ylabel("Grado de «alto»")
plt.ylim(-0.05, 1.1)
plt.title("Gráfica 7 · La rampa de 160 a 200 cm")
plt.show()
````

## Paso 10 · La cuenta, con Juan como ejemplo

¿Cuánta batería tiene Juan (175 cm)? Se hace en **tres pasos**:

1. **¿Cuánto mide la rampa completa?** De 160 a 200: **200 − 160 = 40 cm**.
2. **¿Cuánto ha subido Juan por la rampa?** De 160 a 175: **175 − 160 = 15 cm**.
3. **Divida lo que subió entre lo que mide la rampa:** **15 ÷ 40 = 0,375**.

Juan es alto en **0,375**: un poco más de un tercio de batería.

```
grado = (lo que subió) ÷ (lo que mide la rampa)
      = (175 − 160) ÷ (200 − 160)
      = 15 ÷ 40
      = 0,375
```

La gráfica muestra esos dos tramos: lo que **subió** (15 cm) y lo que **mide la rampa** (40 cm).

````python
juan = 175
plt.figure(figsize=(9, 4.2))
plt.plot(xs, [grado(x) for x in xs], color=NARANJA, linewidth=4)
plt.annotate("", xy=(FINAL, -0.12), xytext=(COMIENZO, -0.12),
             arrowprops=dict(arrowstyle="<->", color="black", lw=1.5))
plt.text(180, -0.2, "la rampa mide 40 cm (200 − 160)", ha="center")
plt.annotate("", xy=(juan, 0.05), xytext=(COMIENZO, 0.05),
             arrowprops=dict(arrowstyle="<->", color="#1f6feb", lw=2.5))
plt.text(167.5, 0.1, "subió 15", ha="center", color="#1f6feb", fontweight="bold")
plt.plot([juan, juan], [0, grado(juan)], color="#1f6feb", linestyle=":", lw=2)
plt.plot([140, juan], [grado(juan), grado(juan)], color="#1f6feb", linestyle=":", lw=2)
plt.scatter(juan, grado(juan), s=150, color="#1f6feb", zorder=3)
plt.text(juan + 2, grado(juan) - 0.04, "Juan: 15 ÷ 40 = 0,375", fontsize=12, fontweight="bold")
plt.xlabel("Estatura (cm)")
plt.ylabel("Grado de «alto»")
plt.ylim(-0.28, 1.1)
plt.xlim(140, 220)
plt.title("Gráfica 8 · Paso a paso: el grado de Juan")
plt.show()
````

## Paso 11 · La misma cuenta para todos

Ahora la celda hace la cuenta de las cinco personas y escribe cada operación, para que pueda repetirla a mano o en Excel.

En **Excel** la fórmula es (si su Excel está en inglés, use `MAX` y `MIN` con comas):

```
=MAX(0;MIN(1;(H3-160)/(200-160)))
```

````python
for p, e in zip(personas, estaturas):
    if e <= COMIENZO:
        print(f"{p:6} {e} cm → está antes de la rampa (160) → grado 0")
    elif e >= FINAL:
        print(f"{p:6} {e} cm → ya pasó la rampa (200) → grado 1")
    else:
        print(f"{p:6} {e} cm → ({e} − 160) ÷ 40 = {e - 160} ÷ 40 = {grado(e):.3f}")
````

## Paso 12 · Las cinco personas sobre la rampa

Cada punto es una persona. Ahora cada una tiene **su propio grado**, según qué tan arriba de la rampa está.

````python
plt.figure(figsize=(9, 4.5))
plt.plot(xs, [grado(x) for x in xs], color=NARANJA, linewidth=4, label="lógica difusa (rampa)")
plt.step(xs, [1 if x >= UMBRAL_ALTO else 0 for x in xs], where="post",
         color=GRIS, linestyle="--", linewidth=2, label="lógica clásica (umbral 180)")
for p, e in zip(personas, estaturas):
    g = grado(e)
    plt.scatter(e, g, s=120, zorder=3, color="#1f6feb")
    plt.text(e + 1.5, g + 0.04, f"{p} {g:.3f}", fontsize=10)
plt.xlabel("Estatura (cm)")
plt.ylabel("Grado de «alto»")
plt.ylim(-0.08, 1.15)
plt.legend(loc="upper left")
plt.grid(alpha=0.3)
plt.title("Gráfica 9 · Rampa (difusa) contra escalón (clásica)")
plt.show()
````

## Paso 13 · Las baterías de cada persona

Esta es la misma información, pero dibujada como **baterías**. La parte naranja es «qué tan alta es» cada persona.

````python
plt.figure(figsize=(9, 4))
for i, (e, p) in enumerate(orden):
    g = grado(e)
    plt.barh(i, 1, color="#eeeeee", edgecolor="black", height=0.6)
    plt.barh(i, g, color=NARANJA, height=0.6)
    plt.text(1.03, i, f"{g * 100:.1f} %  ({g:.3f})", va="center")
plt.yticks(range(len(orden)), [f"{p} ({e} cm)" for e, p in orden])
plt.xlim(0, 1.35)
plt.xticks([0, 0.25, 0.5, 0.75, 1], ["0", "0,25", "0,5", "0,75", "1"])
plt.title("Gráfica 10 · ¿Cuánta «batería de alto» tiene cada uno?")
plt.show()
````

## Paso 14 · Antes y después, lado a lado

Las barras **grises** son la lógica clásica (solo 0 o 1). Las **naranjas** son la lógica difusa (cualquier valor entre 0 y 1).

````python
import numpy as np

nombres = [p for _, p in orden]
c_vals = [1 if e >= UMBRAL_ALTO else 0 for e, _ in orden]
d_vals = [grado(e) for e, _ in orden]
pos = np.arange(len(nombres))

plt.figure(figsize=(9, 4.2))
plt.bar(pos - 0.2, c_vals, width=0.4, color=GRIS, label="Clásica (0 o 1)")
plt.bar(pos + 0.2, d_vals, width=0.4, color=NARANJA, label="Difusa (de 0 a 1)")
for i in pos:
    plt.text(i - 0.2, c_vals[i] + 0.03, f"{c_vals[i]}", ha="center")
    plt.text(i + 0.2, d_vals[i] + 0.03, f"{d_vals[i]:.3f}", ha="center")
plt.xticks(pos, nombres)
plt.ylim(0, 1.25)
plt.ylabel("¿Es alto?")
plt.legend(loc="upper left")
plt.title("Gráfica 11 · Lo mismo con las dos lógicas")
plt.show()

pd.DataFrame({"Persona": personas, "Estatura (cm)": estaturas,
              "Clásica (A: 180)": clasica,
              "Difusa (rampa 160-200)": [round(grado(e), 3) for e in estaturas]})
````

## Paso 16 · ¿Y las notas con lógica difusa?

Se puede hacer, **solo como ejercicio**, con la frase «**le fue bien**» y una rampa de 2,5 (nada bien) a 4,5 (muy bien). Así, 3,0 y 2,99 quedan casi iguales (0,25 y 0,245), como debe ser. Pero el **resultado oficial** sigue siendo pasó o perdió, porque esa decisión sí es de sí o no.

## Paso 17 · Andrés y Bruno: el ejemplo de la presentación

La presentación usa la rampa de **160 a 190 cm** (no la de 160 a 200 de la parte B). Con esa rampa:

- **Andrés** mide **1,80 m** → (180 − 160) ÷ (190 − 160) = 20 ÷ 30 = **0,667**.
- **Bruno** mide **1,79 m** → (179 − 160) ÷ 30 = 19 ÷ 30 = **0,633**.

Con lógica clásica (umbral 180), Andrés es **1** y Bruno es **0**. Con lógica difusa quedan **casi iguales**, como debe ser: el cambio ya no es un salto, es una rampa.

````python
RAMPA_PRES = (160, 190)   # la rampa de la presentación

for nombre, cm in [("Andrés", 180), ("Bruno", 179)]:
    g = grado(cm, *RAMPA_PRES)
    clas = 1 if cm >= 180 else 0
    print(f"{nombre:7} {cm} cm → clásica: {clas} | difusa: ({cm} − 160) ÷ 30 = {cm - 160} ÷ 30 = {g:.3f}")

xs2 = [x / 2 for x in range(300, 401)]   # de 150 a 200 cm, de medio en medio
plt.figure(figsize=(9, 4))
plt.plot(xs2, [grado(x, *RAMPA_PRES) for x in xs2], color=NARANJA, linewidth=4, label="difusa (rampa 160-190)")
plt.step(xs2, [1 if x >= 180 else 0 for x in xs2], where="post", color=GRIS, linestyle="--",
         linewidth=2, label="clásica (umbral 180)")
for nombre, cm, dy in [("Andrés", 180, 0.06), ("Bruno", 179, -0.1)]:
    g = grado(cm, *RAMPA_PRES)
    plt.scatter(cm, g, s=120, color="#1f6feb", zorder=3)
    plt.text(cm - 9, g + dy, f"{nombre} {cm} cm → {g:.3f}", fontsize=10)
plt.xlabel("Estatura (cm)")
plt.ylabel("Grado de «alto»")
plt.legend(loc="upper left")
plt.grid(alpha=0.3)
plt.title("Gráfica 14 · Andrés y Bruno: con la rampa quedan casi iguales")
plt.show()
````

## Paso 18 · La fórmula de la rampa, en una sola línea

Todo lo que hicimos con Juan se resume en una fórmula:

```
grado = (valor − comienzo) ÷ (final − comienzo)
```

con dos reglas extra:

- Si el valor está **antes del comienzo**, el grado es **0**.
- Si el valor está **después del final**, el grado es **1**.

**Cuidado:** la fórmula solo se usa **dentro** de la rampa. Para 1,95 m con la rampa 160-190, la fórmula daría 35 ÷ 30 = 1,17, que **no tiene sentido** (el grado nunca pasa de 1). Por eso, si ya pasó el final, el grado es simplemente 1.

## Paso 19 · Los ejemplos de la diapositiva «Practique con otras estaturas»

Con la rampa **160-190** de la presentación:

````python
ejemplos = [160, 172, 175, 184, 195]
filas = []
for cm in ejemplos:
    g = grado(cm, *RAMPA_PRES)
    if cm <= 160:
        cuenta = "está en el comienzo o antes → 0"
    elif cm >= 190:
        cuenta = "ya pasó el final → 1"
    else:
        cuenta = f"({cm} − 160) ÷ 30 = {cm - 160} ÷ 30"
    filas.append({"Estatura (cm)": cm, "Cuenta": cuenta, "Grado": round(g, 3)})
pd.DataFrame(filas)
````

Fíjese en **172 cm → 0,4**: es justo la **pregunta 2 del quiz** de la presentación. Y en **175 cm → 0,5**: está exactamente en la mitad de la rampa.

## Paso 20 · Varios conjuntos a la vez: frío, templado y caliente

Hasta ahora usamos **una sola** palabra («alto»). Para controlar un **aire acondicionado** se usan **tres** palabras al mismo tiempo, cada una con su propia forma:

| Conjunto | Forma | Vale 1 | Vale 0 |
|---|---|---|---|
| **Frío** | rampa que **baja** | hasta 15 °C | desde 20 °C |
| **Templado** | **triángulo** (sube y baja) | justo en 20 °C | antes de 15 °C y después de 25 °C |
| **Caliente** | rampa que **sube** | desde 30 °C | hasta 20 °C |

Ejecute la celda para verlos juntos en una sola gráfica.

````python
def frio(t):
    if t <= 15: return 1.0
    if t >= 20: return 0.0
    return (20 - t) / 5            # rampa que baja: cuánto FALTA para llegar a 20

def templado(t):
    if t <= 15 or t >= 25: return 0.0
    if t <= 20: return (t - 15) / 5   # subida del triángulo: cuánto AVANZÓ desde 15
    return (25 - t) / 5               # bajada del triángulo: cuánto FALTA para llegar a 25

def caliente(t):
    if t <= 20: return 0.0
    if t >= 30: return 1.0
    return (t - 20) / 10           # rampa que sube: cuánto AVANZÓ desde 20

ts = [x / 10 for x in range(100, 351)]   # de 10 a 35 °C
AZUL = "#1f77b4"

plt.figure(figsize=(9, 4))
plt.plot(ts, [frio(t) for t in ts], color=AZUL, linewidth=3, label="frío")
plt.plot(ts, [templado(t) for t in ts], color=VERDE, linewidth=3, label="templado")
plt.plot(ts, [caliente(t) for t in ts], color=NARANJA, linewidth=3, label="caliente")
plt.xlabel("Temperatura (°C)")
plt.ylabel("Grado")
plt.ylim(-0.05, 1.15)
plt.legend(loc="center right")
plt.grid(alpha=0.3)
plt.title("Gráfica 15 · Tres conjuntos difusos para la temperatura")
plt.show()
````

## Paso 21 · ¿Qué pasa a 22 °C?

Trace una línea vertical en **22 °C**. Esa línea **corta dos curvas** a la vez:

- **Templado:** 22 está en la **bajada** del triángulo (de 20 a 25). Faltan **3** grados para llegar a 25, de una bajada de **5**. Grado = 3 ÷ 5 = **0,6**.
- **Caliente:** 22 está en la **subida** (de 20 a 30). Ha avanzado **2** grados de una subida de **10**. Grado = 2 ÷ 10 = **0,2**.
- **Frío:** 22 ya pasó los 20, así que el grado es **0**.

En palabras: a 22 °C está **bastante templado y un poco caliente al mismo tiempo**. «Está agradable, pero ya empieza a hacer calorcito».

````python
T = 22
plt.figure(figsize=(9, 4.2))
plt.plot(ts, [frio(t) for t in ts], color=AZUL, linewidth=3, label="frío")
plt.plot(ts, [templado(t) for t in ts], color=VERDE, linewidth=3, label="templado")
plt.plot(ts, [caliente(t) for t in ts], color=NARANJA, linewidth=3, label="caliente")
plt.axvline(T, color="black", linestyle="--")
for f, col, nombre in [(templado, VERDE, "templado"), (caliente, NARANJA, "caliente"), (frio, AZUL, "frío")]:
    g = f(T)
    plt.scatter(T, g, s=120, color=col, zorder=3, edgecolor="black")
    plt.text(T + 0.4, g + 0.03, f"{nombre} = {g:.1f}", fontsize=11, fontweight="bold", color=col)
plt.xlabel("Temperatura (°C)")
plt.ylabel("Grado")
plt.ylim(-0.08, 1.15)
plt.legend(loc="upper right")
plt.grid(alpha=0.3)
plt.title("Gráfica 16 · A 22 °C: templado 0,6, caliente 0,2 y frío 0")
plt.show()
````

## Paso 24 · Las reglas difusas del aire acondicionado

Ahora usamos los grados para **decidir**. Las reglas se escriben casi como se hablan:

| Regla | Si hace… | El aire trabaja al… |
|---|---|---|
| Regla 1 | **templado** | **30 %** (enfría suave) |
| Regla 2 | **caliente** | **90 %** (enfría fuerte) |

En la lógica clásica, una regla se cumple o no se cumple. En la lógica difusa, **cada regla se cumple en el mismo grado que su condición**. A 22 °C:

- La regla 1 se cumple en **0,6** (porque está templado en 0,6).
- La regla 2 se cumple en **0,2** (porque está caliente en 0,2).

## Paso 25 · El promedio ponderado: la misma cuenta de la nota definitiva

¿Qué potencia elige el aire? Promedia lo que recomienda cada regla, **dándole más peso a la que más se cumple**. Es igual que calcular una definitiva donde un parcial vale más que otro:

```
potencia = (0,6 × 30 + 0,2 × 90) ÷ (0,6 + 0,2)
         = (18 + 18) ÷ 0,8
         = 36 ÷ 0,8
         = 45 %
```

A 22 °C el aire trabaja al **45 %**: más que el modo suave (30 %), porque ya empieza a hacer calor, pero lejos del modo fuerte (90 %).

````python
def potencia(t, suave=30, fuerte=90):
    g1, g2 = templado(t), caliente(t)
    if g1 + g2 == 0:
        return 0.0, g1, g2          # ninguna regla se cumple: el aire no enfría
    return (g1 * suave + g2 * fuerte) / (g1 + g2), g1, g2

for t in [22, 27]:
    p, g1, g2 = potencia(t)
    print(f"{t} °C → templado {g1:.1f}, caliente {g2:.1f} → "
          f"({g1:.1f} × 30 + {g2:.1f} × 90) ÷ ({g1:.1f} + {g2:.1f}) = {p:.0f} %")

p, g1, g2 = potencia(22)
fig, ax = plt.subplots(figsize=(9, 3.2))
ax.barh(["Regla 2 · caliente → 90 %", "Regla 1 · templado → 30 %"], [g2, g1], color=[NARANJA, VERDE])
ax.text(g2 + 0.01, 0, f"peso {g2:.1f}", va="center")
ax.text(g1 + 0.01, 1, f"peso {g1:.1f}", va="center")
ax.set_xlim(0, 1)
ax.set_xlabel("Qué tanto se cumple cada regla a 22 °C")
ax.set_title(f"Gráfica 18 · A 22 °C pesa más la regla 1, y el resultado es {p:.0f} %")
plt.show()
````

**A 27 °C:** templado baja a **0** y caliente sube a **0,7**. Solo queda activa la regla 2, así que el aire trabaja al **90 %**.

## Paso 27 · Termostato clásico contra aire difuso

Un termostato clásico usa un **umbral**: «si pasa de 24 °C, a tope (100 %); si no, apagado (0 %)». Compárelo con el aire difuso:
