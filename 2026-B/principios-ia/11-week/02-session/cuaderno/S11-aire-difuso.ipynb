Este cuaderno es el aire acondicionado de la clase: tres conjuntos difusos (**frío**, **templado** y **caliente**) y dos reglas. Vas a ver cómo decide su potencia de forma gradual, y qué pasa cuando cambias dónde empieza «caliente».

```python
# 1. Traer las herramientas (se ejecuta una sola vez)
import numpy as np
import matplotlib.pyplot as plt
print("Listo: herramientas cargadas.")
```

## 1. Los tres conjuntos difusos

Son las mismas rampas y el mismo triángulo de la clase. El único valor que puedes mover aquí es **dónde empieza «caliente»**. Déjalo primero en 20, como en la clase.

```python
# 2. Definir los conjuntos y dibujarlos
caliente_empieza = 20   # ⬅️ CAMBIA ESTE NÚMERO (solo en la parte 3 de la guía)

def frio(t):      # 1 hasta 15 grados, baja a 0 en 20
    return float(np.clip((20 - t) / 5, 0, 1))
def templado(t):  # triangulo: 0 en 15, 1 en 20, 0 en 25
    return float(max(0, min((t - 15) / 5, (25 - t) / 5)))
def caliente(t):  # 0 hasta caliente_empieza, sube a 1 diez grados despues
    return float(np.clip((t - caliente_empieza) / 10, 0, 1))

temps = np.linspace(10, 35, 251)
plt.plot(temps, [frio(t) for t in temps], label="frío")
plt.plot(temps, [templado(t) for t in temps], label="templado")
plt.plot(temps, [caliente(t) for t in temps], label="caliente")
plt.xlabel("temperatura (°C)")
plt.ylabel("grado de pertenencia")
plt.title("Los tres conjuntos difusos")
plt.legend()
plt.show()
```

## 2. ¿Qué decide el aire a cierta temperatura?

Las reglas son las de la clase:

- si hace **templado**, enfría suave: **30%** de potencia;
- si hace **caliente**, enfría fuerte: **90%** de potencia.

Cada regla pesa según el grado de su condición, y la potencia final es el promedio ponderado. Empieza con **22 grados**.

```python
# 3. Calcular los grados y la potencia
temperatura = 22   # ⬅️ CAMBIA ESTE NÚMERO: prueba 12, 18, 22, 27 y 32

g_frio, g_templado, g_caliente = frio(temperatura), templado(temperatura), caliente(temperatura)
print(f"A {temperatura} °C:")
print(f"  frío     = {g_frio:.2f}")
print(f"  templado = {g_templado:.2f}")
print(f"  caliente = {g_caliente:.2f}")

peso_total = g_templado + g_caliente
if peso_total == 0:
    potencia = 0          # si solo hace frio, el aire se apaga
else:
    potencia = (g_templado * 30 + g_caliente * 90) / peso_total
print(f"  -> el aire trabaja al {potencia:.0f}% de su potencia")
```

## 3. Difuso contra sí o no

Esta gráfica compara el aire difuso con un termostato clásico que solo sabe prender a tope o apagar, con el corte en 24 °C.

```python
# 4. Comparar las dos formas de decidir en todas las temperaturas
def potencia_difusa(t):
    gt, gc = templado(t), caliente(t)
    return 0 if gt + gc == 0 else (gt * 30 + gc * 90) / (gt + gc)

plt.plot(temps, [potencia_difusa(t) for t in temps], label="aire difuso")
plt.plot(temps, [100 if t >= 24 else 0 for t in temps], label="termostato de sí o no")
plt.xlabel("temperatura (°C)")
plt.ylabel("potencia (%)")
plt.title("Dos formas de decidir la potencia")
plt.legend()
plt.show()
```
